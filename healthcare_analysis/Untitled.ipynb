```python
class BankAccount:
    def __init__ (self, name, balance=0):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"Hey {self.name}, your balance is {self.balance}"

    def withdraw(self, amount):
        if amount > self.balance:
            return "Low balance"
        else:
            self.balance -= amount
            return f"You withdrew KES {amount}, your balance is {self.balance}"
        
        
    
```

```python
bank1 = BankAccount("Peter")
print(bank1.deposit(50000))

print(bank1.deposit(40000))

```
